```python
# Função para o Redimensionamento de uma Imagem
# Imagem de Gato

from PIL import Image

#Conversão da imagem
def converter_para_cinza(imagem):
    imagem_cinza = imagem.convert('L')
    return imagem_cinza


def binarizar_imagem(imagem_cinza, limiar=127):
    imagem_binaria = imagem_cinza.point(lambda p: p > limiar and 255)
    return imagem_binaria

# Abrir a imagem
imagem = Image.open('/content/Cat (1008).jpg')

# Converte para tons de cinza
imagem_cinza = converter_para_cinza(imagem)

# Binariza a imagem
imagem_binaria = binarizar_imagem(imagem_cinza)

display(imagem), display(imagem_cinza), display(imagem_binaria)
```
